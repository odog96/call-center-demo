```python
# Distribution of call types
query2 = """
SELECT 
    call_type, 
    COUNT(*) as count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM calls), 2) as percentage
FROM calls
GROUP BY call_type
"""
pd.read_sql_query(query2, conn)
```

```python
# Average sentiment by profile type
query3 = """
SELECT 
    customer_profile_type,
    ROUND(AVG(overall_sentiment), 2) as avg_sentiment,
    COUNT(*) as call_count
FROM calls
GROUP BY customer_profile_type
ORDER BY avg_sentiment DESC
"""
pd.read_sql_query(query3, conn)

```
