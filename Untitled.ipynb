```python
import json
import pandas as pd

# For each JSON file
for filename in ['call_log.json', 'customers.json', 'promotions.json']:  # replace with your actual filenames
    with open(filename, 'r') as f:
        data = json.load(f)
    
    # Convert to dataframe and save as CSV
    pd.DataFrame(data).to_csv(filename.replace('.json', '.csv'), index=False)
```
